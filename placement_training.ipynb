```python
r = int(input("Enter number of rats: "))
unit = int(input("Enter amount of food each rat eats: "))
n = int(input("Enter number of houses: "))
arr = list(map(int, input("Enter food in each house: ").split()))
if arr is None:
    print(-1)
else:
    required = r * unit
    total = 0
    houses = 0
    for i in range(n):
        total = total + arr[i]
        houses = houses + 1
        if total >= required:
            break
    if total < required:
        print(0)
    else:
        print("Minimum number of houses:", houses)
```

```python
n = int(input("Enter a number: "))
largest = 0
while n > 0:
    digit = n % 10
    if digit > largest:
        largest = digit
    n = n // 10
print("Largest digit:", largest)
```
