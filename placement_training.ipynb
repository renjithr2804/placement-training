```python
n=int(input("Enter n:"))
k=int(input("Enter k:"))
j=int(input("Enter j:"))
m=int(input("Enter m:"))
p=int(input("Enter p:"))
if(k==0 or j==0):
  print("INVALID INPUT")
else:
  n=n-(m//k+p//j)
  if(m%k!=0 or p%k!=0):
    n=n-1
  print("No 1of monkeys in tree:",n)
```
